### 실습 1. Excel 파일 불러오기

해야 할 일

pandas를 이용해 Excel 파일을 DataFrame으로 불러옵니다.

```python
import pandas as pd

df = pd.read_excel("교보문고_온라인_베스트셀러_상품리스트.xlsx")
df.head()
```

### 실습 2. 데이터의 크기와 컬럼 확인하기

Excel 파일을 불러왔으면 바로 전처리를 시작하지 않습니다.

먼저 어떤 데이터인지 확인합니다.

```python
# 전체 행과 컬럼 개수
print("행, 컬럼 개수:", df.shape)

# 컬럼 이름 확인
print("=====\n컬럼 이름:=====")
print(df.columns)

# 각 컬럼의 자료형 확인
print("\n자료형:")
print(df.dtypes)

# 각 컬럼의 결측치 개수 확인
print("\n결측치 개수:")
print(df.isna().sum())

# 앞의 5행 확인
print("\n앞의 5행:")
display(df.head())
```

- 나머지 컬럼은 문자열형(`str`)으로 확인되었다.

### 자료형 특이사항

- `정가`, `판매가`, `적립예정포인트`는 숫자처럼 보이지만 현재 `str` 자료형이다.
- `할인율`, `적립율`도 `%`가 포함되어 있어 문자열형으로 저장되어 있다.
- 이후 계산이나 분석을 하려면 이러한 컬럼은 숫자형으로 변환할 필요가 있다.
- `발행(출시)일자`는 날짜처럼 보이지만 현재 `int64`이므로 필요하면 날짜형으로 변환할 수 있다.

```python
df.info()
df.isna().sum().sum()
```

### 실습 3. 분석에 필요한 컬럼만 선택하기

원본 데이터의 모든 컬럼이 이후 분석에 필요한 것은 아닙니다.
이번 과정에서는 다음 컬럼을 먼저 선택합니다.

```python
# 필요한 컬럼만 선택
df_books = df[
    [
        "순위",
        "판매상품 ID",
        "상품명",
        "판매가",
        "인물",
        "출판사",
        "발행(출시)일자",
        "분야"
    ]
]

# 선택한 컬럼 확인
print(df_books.columns)

# 앞의 5행 확인
df_books.head()
```

### 실행 결과

- 필요한 8개 컬럼만 선택하여 `df_books` DataFrame을 만들었다.
- 선택된 컬럼은 `순위`, `판매상품 ID`, `상품명`, `판매가`, `인물`, `출판사`, `발행(출시)일자`, `분야`이다.
- `df_books.columns`로 컬럼이 정상적으로 선택된 것을 확인했다.
- `df_books.head()`를 통해 앞의 5행 데이터도 정상적으로 조회되었다.

### 실습 4. 컬럼 이름을 사용하기 쉽게 정리하기

원본 컬럼 이름은 사람이 읽기에는 좋지만 코드에서 반복해서 사용하기에는 조금 길 수 있습니다.
이번 실습에서는 다음처럼 간단하게 정리할 수 있습니다.

```python
# 컬럼 이름 변경
df_books = df_books.rename(
    columns={
        "판매상품 ID": "판매상품ID",
        "인물": "저자",
        "발행(출시)일자": "발행일"
    }
)

# 변경된 컬럼 이름 확인
print(df_books.columns)
df_books.head()
```

### 실행 결과

- `rename()`을 사용해 컬럼 이름을 더 간단하게 변경했다.
- `판매상품 ID` → `판매상품ID`
- `인물` → `저자`
- `발행(출시)일자` → `발행일`
- `df_books.columns`를 통해 변경된 컬럼명이 정상적으로 적용된 것을 확인했다.

### 실습 5. 결측치 확인하기

결측치는 값이 비어 있는 데이터를 의미합니다.

데이터 분석에서는 결측치를 무조건 삭제하지 않습니다.

먼저 어디에 얼마나 있는지 확인하고, 분석 목적에 맞게 처리 방법을 결정합니다.

```python
# 컬럼별 결측치 개수 확인
print(df_books.isna().sum())

# 결측치가 있는 컬럼만 확인
missing = df_books.isna().sum()
print(missing[missing > 0])
```

### 실행 결과

- `df_books.isna().sum()`으로 각 컬럼의 결측치 개수를 확인했다.
- 모든 컬럼의 결측치 개수가 **0개**로 확인되었다.
- 결측치가 있는 컬럼만 조회한 결과도 비어 있으므로, 현재 `df_books`에는 결측치가 없다.

### 실습 6. 판매가를 숫자로 바꾸기
 
Excel에서 판매가가 다음과 같이 저장되어 있을 수 있습니다.

```python
# 1. 쉼표 제거
df_books["판매가"] = df_books["판매가"].str.replace(",", "")

# 2. 숫자형으로 변환
df_books["판매가"] = df_books["판매가"].astype(int)

# 3. 자료형 확인
print(df_books["판매가"].dtype)
```

### 실행 결과

- `판매가` 컬럼의 쉼표를 제거했다.
- 문자열 형태의 가격을 정수형으로 변환했다.
- 변환 후 자료형이 `int64`로 확인되었다.
- 이제 `판매가` 컬럼을 이용해 평균, 합계 등의 숫자 계산이 가능하다.

### 실습 7. 발행일을 날짜형으로 바꾸기

원본 데이터의 발행일은 다음처럼 보일 수 있습니다.

```python
# 1. 발행일을 날짜형으로 변환
df_books["발행일"] = pd.to_datetime(
    df_books["발행일"],
    format="%Y%m%d"
)

# 2. 자료형 확인
print(df_books["발행일"].dtype)

# 3. 앞의 5개 값 확인
print(df_books["발행일"].head())
```

### 실행 결과

- `발행일` 컬럼을 `pd.to_datetime()`으로 날짜형으로 변환했다.
- `format="%Y%m%d"`를 지정해 `YYYYMMDD` 형식으로 해석했다.
- 변환 후 자료형은 `datetime64[us]`로 확인되었다.
- 앞의 5개 값도 `2014-05-19`처럼 날짜 형식으로 정상 변환되었다.

### 실습 8. 문자열 앞뒤 공백 정리하기

상품명, 저자, 출판사, 분야는 문자열 데이터입니다.
 
문자열 앞뒤에 불필요한 공백이 있으면 같은 값도 서로 다른 데이터처럼 처리될 수 있습니다.

예를 들어 다음 두 값은 사람이 보기에는 같지만 문자열로는 다릅니다.


```python
# 공백을 제거할 컬럼 목록
text_columns = ["상품명", "저자", "출판사", "분야"]

# 각 컬럼의 앞뒤 공백 제거
for column in text_columns:
    df_books[column] = df_books[column].str.strip()

# 처리 결과 확인
for column in text_columns:
    print(f"\n{column}:")
    print(df_books[column].head())
```

### 실행 결과

- `상품명`, `저자`, `출판사`, `분야` 컬럼의 앞뒤 공백을 제거했다.
- `for` 반복문을 사용해 여러 문자열 컬럼에 같은 작업을 적용했다.
- `.str.strip()`을 사용해 불필요한 공백을 정리했다.
- 각 컬럼의 앞 5개 값을 확인해 정상적으로 처리된 것을 확인했다.

### 실습 9. 중복 데이터 확인하기

데이터에서 중복을 발견했다고 해서 바로 삭제하면 안 됩니다.

먼저 무엇을 기준으로 중복인지 생각해야 합니다.

예를 들어 같은 상품명이 여러 번 등장할 수 있습니다.

하지만 같은 제목이라도 다른 판본, 다른 상품, 개정판 등이 있을 수 있습니다.

따라서 이번 데이터에서는 상품명만 보고 중복 행을 바로 삭제하지 않습니다.

우선 상품을 구분하는 판매상품ID를 확인합니다.

```python
# 1. 전체 행이 완전히 같은 중복 개수
print("전체 중복 행:", df_books.duplicated().sum())

# 2. 판매상품ID가 중복된 행 개수
print("판매상품ID 중복:", df_books["판매상품ID"].duplicated().sum())

# 3. 상품명이 중복된 행 개수
print("상품명 중복:", df_books["상품명"].duplicated().sum())
```

### 실행 결과

- 전체 행이 완전히 같은 중복 데이터는 **0개**이다.
- `판매상품ID` 기준 중복도 **0개**이다.
- `상품명` 기준 중복도 **0개**이다.
- 따라서 현재 `df_books`에서는 확인한 기준에 따른 중복 데이터가 없다.
