```python
import pandas as pd

orders = pd.read_csv("../data/raw/orders.csv")
items = pd.read_csv("../data/raw/order_items.csv")

orders["order_date"] = pd.to_datetime(orders["order_date"])

recent = orders[orders["order_status"] == "completed"]
recent = recent[recent["order_date"] >= "2026-03-04"]

items_recent = items[items["order_id"].isin(recent["order_id"])]

result = items_recent.groupby("product_id")["quantity"].sum()
result = result.sort_values(ascending=False)

result.head(10)
```
