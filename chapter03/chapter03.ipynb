### 결과 관찰

범주형 데이터에서는 customers.city 중 성남이 21건으로 가장 많았고, products.category에서는 스포츠가 19건으로 가장 많았다.

### 나의 해석과 판단

현재 확인한 숫자형 데이터의 최소값과 최대값은 눈에 띄게 비정상적으로 보이는 값은 없었다.

범주형 데이터는 값별 빈도 차이가 있었지만, 현재 단계에서는 단순한 편중인지 실제 데이터 특성인지 추가 확인이 필요하다고 판단했다.

order_date는 문자열 상태로 읽혔지만 날짜 타입 변환에 실패한 값이 없었기 때문에 현재 샘플 데이터에서는 날짜 형식이 비교적 일관되게 저장되어 있다고 판단했다.

### 업무·분석적 의미

숫자형 데이터의 범위를 미리 확인하면 잘못 입력된 값이나 비정상적인 값이 존재하는지 빠르게 확인할 수 있다.

범주형 데이터의 분포를 확인하면 특정 지역, 상품군, 주문 상태에 데이터가 지나치게 몰려 있는지 파악할 수 있다.

또한 날짜 데이터를 실제 날짜 타입으로 변환해 두면 이후 기간별 주문 분석이나 시계열 분석을 수행하기 쉬워진다.

### 한계와 추가 확인 사항

현재는 기본적인 범위와 분포만 확인했기 때문에 실제 업무 규칙에 맞는 값인지까지 판단할 수는 없다.

이상값으로 보이는 값이 발견될 경우 임의로 수정하거나 삭제하기보다 실제 업무에서 가능한 값인지 추가 정보와 함께 확인할 필요가 있다.

![숫자형·범주형·날짜 데이터 확인](images/step03_distribution.png)


```python
missing_customer_ids = orders.loc[
    ~orders["customer_id"].isin(customers["customer_id"]),
    "customer_id"
].unique()

missing_order_ids = order_items.loc[
    ~order_items["order_id"].isin(orders["order_id"]),
    "order_id"
].unique()

missing_product_ids = order_items.loc[
    ~order_items["product_id"].isin(products["product_id"]),
    "product_id"
].unique()

print("customers에 없는 customer_id:", len(missing_customer_ids))
print("orders에 없는 order_id:", len(missing_order_ids))
print("products에 없는 product_id:", len(missing_product_ids))
```

## STEP 7 CSV 간 키 관계 검증

### 결과 관찰

orders의 customer_id 중 customers에 존재하지 않는 값은 0개였다.

order_items의 order_id 중 orders에 존재하지 않는 값은 0개였고, order_items의 product_id 중 products에 존재하지 않는 값도 0개였다.

### 나의 해석과 판단

현재 샘플 데이터에서는 자식 테이블에 있는 연결용 ID가 모두 부모 테이블에 존재하고 있어 기본적인 키 관계는 정상적으로 연결되어 있다고 판단했다.

즉, 현재 확인한 범위에서는 orphan key가 발견되지 않았다.

### 업무·분석적 의미

테이블 간 연결용 키가 정상적으로 존재해야 이후 merge나 join을 수행할 때 데이터가 누락되거나 잘못 연결되는 문제를 줄일 수 있다.

여러 CSV를 함께 분석할 경우 개별 테이블의 ID 고유성뿐 아니라 테이블 사이의 연결 관계도 함께 확인하는 것이 중요하다.

### 한계와 추가 확인 사항

현재는 연결 대상 ID가 존재하는지만 확인했기 때문에 각 데이터 값 자체의 정확성이나 실제 업무 규칙까지 검증한 것은 아니다.

추후 실제 운영 데이터에서는 키 생성 규칙이나 데이터 변경 이력 등도 추가로 확인할 필요가 있다.

## STEP 8 LLM을 활용한 추가 데이터 품질 점검

### LLM에 제공한 Safe Context
원본 고객 데이터나 개인정보는 제공하지 않고, 각 데이터셋의 shape, 컬럼명, 데이터 타입, 결측 및 중복 여부, 숫자형 범위, 범주형 분포, 키 관계 검증 결과만 제공했다.

### LLM이 제안한 추가 점검 항목
LLM은 다음과 같은 항목을 추가로 확인할 것을 제안했다.

- signup_date 날짜 변환 가능 여부
- 숫자형 값의 이상 범위 확인
- 범주형 값의 표기 일관성
- 주문 상태별 분포 확인
- products.price와 order_items.unit_price의 관계 확인
- 고객 가입일과 주문일의 관계 확인
- 취소 및 환불 처리 기준 확인

### 실제 데이터에서 확인 가능한 내용
현재 데이터만으로 signup_date의 날짜 변환 여부, 숫자형 값의 범위, 범주형 값의 종류와 빈도, products.price와 order_items.unit_price의 차이 등을 추가로 확인할 수 있다.

반면 나이와 가격이 실제 업무 규칙상 정상 범위인지, 취소 및 환불 상태의 정확한 업무 기준이 무엇인지는 현재 데이터만으로 단정하기 어렵다.

### 채택 / 수정 / 보류한 내용
- 채택: signup_date 날짜 변환 여부, 숫자형 값 범위, 범주형 값 표기 일관성 확인
- 수정: 가격 차이는 단순 오류로 판단하지 않고 할인이나 가격 변경 가능성을 함께 고려
- 보류: 취소 및 환불 처리 기준, 고객 연령 기준 등 업무 규칙이 필요한 항목

### 나의 판단
LLM의 제안은 추가 점검 항목을 찾는 데 도움이 되었지만, 모든 제안을 바로 문제로 판단할 수는 없었다.

특히 데이터에서 직접 확인할 수 있는 사실과 실제 업무 정의가 필요한 내용을 구분해야 한다고 판단했다.

이번 Chapter에서는 데이터를 바로 수정하기보다 현재 상태를 관찰하고 기록하는 것이 목적이므로, 이상해 보이는 값이 있더라도 원인을 확인하기 전에는 임의로 삭제하거나 수정하지 않는 것이 적절하다고 판단했다.

또한 가격 차이가 발견되더라도 할인, 행사, 가격 변경 등 업무상의 이유가 있을 수 있으므로 추가 정보 없이 오류라고 단정하면 안 된다.
