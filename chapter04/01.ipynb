```python
# 현재 폴더를 확인한다.(1)
current = Path.cwd().resolve()
project_root = current
print("현재 폴더 :",current)

# 현재 폴더 안에 data 폴더가 있는지 확인한다.(2)
while True:

# 만약 data 폴더가 있다면
    if (current / "data").is_dir():
#   현재 폴더를 프로젝트 루트로 결정한다.
        project_root = current
#   탐색을 종료한다.
        break

# 만약 data 폴더가 없다면 
#   부모 폴더로 한 단계 올라간다.
    current = current.parent

# 다시 data 폴더가 있는지 확인한다.
# 이 과정을 반복한다.
 
# 더 이상 위로 올라갈 수 없는데도
 
# 데이터 폴더를 찾지 못했다면
 
# 오류를 발생시킨다.

print("프로젝트 루트 :",project_root)
```

```python
from pathlib import Path

def get_project_root():
    current = Path.cwd().resolve()
    project_root = current

    while True:
        if (current / "data").is_dir():
            project_root = current
            break

        current = current.parent

    return project_root

def get_data_dir():
    return get_project_root() / "data"

# get_data_dir() 이라는 함수를 만드시오.
```
